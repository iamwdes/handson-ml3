## 先决条件

### 需了解
* **Python** – 您不需要成为专业Python程序员，但是您确实需要了解基础知识。如果不这样做, 官方[Python教程](https://docs.python.org/3/tutorial/) 是一个不错的起点。
* **Scientific Python** – 我们将使用一些流行的Python库，特别是Numpy，Matplotlib和Pandas。如果您不熟悉这些库，则可能应该首先浏览“工具”部分中的教程（尤其是numpy）。
* **Math** – 我们还将使用一些线性代数，微积分，统计和概率理论的概念。如果您过去学会了这些内容，您应该能够跟随它，因为它不会很高深，但是如果您不知道这些主题或需要进修，那么请仔细阅读数学部分中的适当介绍。

### 运行示例
* **Jupyter** – 这些笔记本基于jupyter。您可以使用托管平台（例如Binder、Deepnote或Colaboratory（无需安装））单击一键即可运行这些笔记本，也可以使用jupyter.org的查看器查看它们，或者您可以在计算机上安装所有内容，因为您喜欢在计算机上安装所有内容。有关更多详细信息，请查看[主页](https://github.com/ageron/handson-ml3/)。
